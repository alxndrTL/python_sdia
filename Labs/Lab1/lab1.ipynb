Given the following list, 

  1. extract the first two items

  4. extract the items located at odd indices
  5. create a new list which corresponds to the reversed version of the originial list

````python
x = [3, 7, 5, 3, 8, 6, 8, 8, 0, 7, 3, 9, 3, 4, 2, 7, 0, 9, 5, 0, 0, 9, 0, 9, 3, 1, 4, 0, 5, 5, 5, 8, 9, 9, 5, 5, 4, 9, 5, 6, 2, 8, 5, 2, 4, 9, 2, 2, 3, 1]

# 1. deux premiers éléments
print("deux premiers éléments:",)
print(x[0:2])

# 2. deux derniers éléments
print("deux derniers éléments:",)
print(x[-2:])

# 3. index de la première occurence de 9
print("index de la première occurence de 9:",)
# méthode la plus simple
for (i, element) in enumerate(x):
    if element == 9:
        print(i)
        break

# méthode avancée avec la fonction index (qui renvoit une erreur si l'élément n'est pas dans la liste)
if 9 in x:
    print(x.index(9))
else:
    print("9 n'est pas dans la liste")

# 4. éléments d'index impairs
print(x[1::2]) # on commence à l'indice 1, et on fait des pas de 2 éléments, ce qui au totla revient à prendre les éléments d'index impairs

# 5. liste inverse
print("liste inverse:",)
# methode simple
reversed1 = [None]*len(x) # liste remplie de None de la même taille que x
for i in range(len(x)):
    # on parcourt x et on remplit reversed1 en partant de la fin
    reversed1[i] = x[len(x)-1-i]
print(reversed1)

# methode plus efficace avec le slicing
reversed2 = x[::-1]
print(reversed2)

# x.reverse() n'est pas approprié car il modifie x en place, il ne renvoit pas de nouvelle liste comme demandé

# 6. liste avec x^3
print("éléments de x à la puissance 3:",)
# methdode simple
cubed1 = []
for element in x:
    cubed1.append(element**3)
print(cubed1)

# plus avancée avec la compréhension de liste
cubed2 = [element**3 for element in x]
print(cubed2)
````

## Functions

Define the function `is_unique(x)` which takes a list of integers `x` as input and returns `True` if the there is no duplicate items otherwise return `False`.

````python
# globalement, on évite les solutions en O(n^2) qui sont du type :
# for i in range(len(x)):
#     for j in range(i + 1, len(x)):
#         if x[i] == x[j]:
#             return False

# on va donc utiliser un set (le check d'appartenance dans un set est en O(1))

# NOTE : cette fonction est reproduite (et documentée prorement avec Docstrings) dans le fichier src/functions.py

def is_unique(x):
    """Indique si une liste ne contient aucun doublon.

    Args:
        x (list): liste d'éléments (entiers, chaînes, etc.).

    Returns:
        bool: ``True`` si tous les éléments de ``x`` sont distincts,
        ``False`` s'il existe au moins un doublon.

    Example:
        >>> is_unique([1, 2, 3])
        True
        >>> is_unique([1, 2, 1])
        False
    """
    return len(set(x)) == len(x)

# ou plus simple encore :

def is_unique(x):
    return len(set(x)) == len(x)
````

````python
print(is_unique([1, 2, 3, 4, 5])) # True
print(is_unique([1, 2, 3, 4, 1])) # False
````

Define a function `triangle_shape(height)`, that returns 
- `""` if `height=0`, 
- otherwise it returns a string that forms a triangle with height prescribed by `height`.

**Examples**

height=1
```
x
```

height=2
```
 x 
xxx
```

height=3
```
  x  
 xxx 
xxxxx
```

height=4
```
   x   
  xxx  
 xxxxx 
xxxxxxx
```

height=5
```
    x    
   xxx   
  xxxxx  
 xxxxxxx 
xxxxxxxxx
```

height=6
```
     x     
    xxx    
   xxxxx   
  xxxxxxx  
 xxxxxxxxx 
xxxxxxxxxxx
```

````python
# voici notre première solution pour comprendre le problème

def triangle_shape(height):
    myStr = ""
    for i in range(1, height + 1):
        myStr += "*" * i + "\n"
    return myStr

#*
#**
#***
#****
#*****

# qui est fausse car elle donne un triangle rectangle, pas une pyramide
# il manque des espaces avant les *

# à la main, on remarque qu'il faut (height - 1 - i) espaces pour la ligne i, d'où:
def triangle_shape(height):
    """Construit un triangle isocèle de caractères.

    Args:
        height (int): hauteur du triangle, entier positif ou nul.

    Returns:
        str: la chaîne représentant le triangle, ``""`` si ``height`` vaut 0.

    Example:
        >>> print(triangle_shape(3))
          x
         xxx
        xxxxx
    """
    result = ""
    for i in range(height):
        spaces = " " * (height - 1 - i)
        result += spaces + "x" * (2 * i + 1) + spaces
        if i < height - 1:
            result += "\n"
    return result
````

````python
print(triangle_shape(5))
````
